```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df = sns.load_dataset('iris')
df.head()
```

Implement advanced plots (box plot, violin plot, pair plot).

```python
# 1) Box plot
plt.figure(figsize=(20,15))

plt.subplot(2,2,1)
sns.boxplot(df)
plt.title("Box plot of species")

# 2) Violin plot
plt.subplot(2,2,2)
sns.violinplot(data=df, x="sepal_width")
plt.title("Violin plot of sepal_width")

# 3) Pair plot
plt.subplot(2,2,3)
sns.pairplot(df)
plt.title("Pairplot of df")

plt.tight_layout()
plt.show()
```

Distribution analysis (skewness, kurtosis).


```python
for col in df.columns:
    skew = df[col].skew
    kurto = df[col].kurt
    print(f"skewness : {skew}, kurtosis : {kurto}")
```

Correlation Analysis

a) Understanding correlation and its importance.

b) Calculating correlation coefficients (Pearson, Spearman).

c) Visualizing correlations (heatmaps)

```python
corr_pearson = df.select_dtypes(['int64', 'float64']).corr(method="pearson")
corr_spearman = df.select_dtypes(['int64', 'float64']).corr(method="spearman")

print(f"Pearson coorelation {corr_pearson}")
print(f"Spearman coorelation {corr_spearman}")
```
