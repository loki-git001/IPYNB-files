```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama

# 1. Create a simple prompt template with a {topic} variable
template = ChatPromptTemplate(
    [
        ("system", "You are a the best comedian. Your name is {name}."),
        ("human", "I bet you can't tell a joke about the topic that can make me laugh a lot hearing that."),
        ("ai", "Let's see. Give me the topic first!"),
        ("human", "Ok then the topic is {topic}")
    ]
)

# 2. Inspect the resulting prompts by supplying different topics
topics = ["tea", "coding", "human", "AI"]
for topic in topics:
    formatted_prompt = template.invoke({"name": "Sasi", "topic":topic})
    print(f"--- Formatted Prompt for '{topic}' ---")
    print(formatted_prompt)

# 3. Initialize the local Llama 3.2 chat model via Ollama
model = ChatOllama(model="llama3.2")

# 4. Combine the prompt template and model into a chain and invoke it
chain = template | model

for topic in topics:
    response = chain.invoke({"name": "Sasi", "topic": topic})
    print(f"--- LLM response for '{topic}' ---")
    print(response.content)
```

```python
for topic in topics:
    print(
        f"\n--------------------------- LLM response for '{topic}' ---------------------------"
    )
    for chunk in chain.stream({"name": "Sasi", "topic": topic}):
        print(chunk.content, end="", flush=True)
```
