```python
import asyncio
from llama_index.core.agent.workflow import FunctionAgent
from llama_index.llms.ollama import Ollama

# Define a simple calculator tool
def multiply(a: float, b: float) -> float:
    """Useful for multiplying two numbers."""
    return a * b

llm = Ollama(
    model="llama3.2",
    request_timeout=120.0,
    # Manually set the context window to limit memory usage
    context_window=2000,
)

# Create an agent workflow with our calculator tool
agent = FunctionAgent(
    tools=[multiply],
    llm=llm,
    system_prompt="You are a helpful assistant that can multiply two numbers. You also answers basic questions.",
)


response = await agent.run("What is 1234 * 4567?")
print(str(response))
```

```python
from llama_index.core.workflow import Context

# create context
ctx = Context(agent)

# run agent with context
response = await agent.run("My name is Logan", ctx=ctx)
print(str(response))
response = await agent.run("What is my name?", ctx=ctx)
print(str(response))
```

```python
from llama_index.core import (
    SimpleDirectoryReader,
    VectorStoreIndex,
    Settings
)
from llama_index.embeddings.ollama import OllamaEmbedding

embed_model = OllamaEmbedding(model_name="nomic-embed-text")
Settings.embed_model = embed_model
Settings.llm = llm

# Loading
documents = SimpleDirectoryReader("data").load_data()

# Indexing
index = VectorStoreIndex.from_documents(documents)

query_engine = index.as_query_engine()

# retrieving
response = query_engine.query("give me the exact text in documents that I gave you to take the context from")

print(response)
```

```python
response = query_engine.query("What is a topic that doesn't appear in the document?")

print(response)
```

```python
response = query_engine.query("what is machine learning")

print(response)
```
