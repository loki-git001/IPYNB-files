# STACK


- **Array Stack**


```python
from array import array


class ArrayStack:
    def __init__(self, dtype="i", values=None, capacity=0):
        if values and capacity == 0:
            capacity = len(values)

        self.capacity = capacity
        self.dtype = dtype
        self.stack = array(dtype, [0] * capacity)
        self.top = -1
        if values:
            for val in values:
                self.push(val)

    def push(self, data):
        if self.is_full():
            raise IndexError("Stack Overflow : cannot push onto a full stack")
        self.top += 1
        self.stack[self.top] = data

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack Underflow : cannot pop from empty stack")
        item = self.stack[self.top]
        self.top -= 1
        return item

    def peek(self):
        if self.is_empty():
            raise IndexError("Stack Underflow : cannot pop from empty stack")
        return self.stack[self.top]

    def is_empty(self):
        return self.top == -1

    def is_full(self):
        return self.top == self.capacity - 1

    def size(self):
        return self.top + 1

    def __repr__(self):
        if self.is_empty():
            return "[ empty stack ]"

        elements = [str(self.stack[i]) for i in range(self.top, -1, -1)]
        body = "  ↑ top\n" + "\n".join(f"| {elem} |" for elem in elements)
        return f"{body} \n(capacity = {self.capacity}, dtype = '{self.dtype}')"


def test_array_stack():
    print("=== Test 1: Create empty stack ===")
    s = ArrayStack(dtype="i", capacity=5)
    print(s)
    assert s.is_empty()
    assert not s.is_full()
    assert s.size() == 0

    print("\n=== Test 2: Push elements ===")
    s.push(10)
    s.push(20)
    s.push(30)
    print(s)
    assert s.size() == 3
    assert s.peek() == 30
    assert not s.is_empty()

    print("\n=== Test 3: Pop elements ===")
    popped = s.pop()
    print(f"Popped value: {popped}")
    assert popped == 30
    assert s.peek() == 20
    assert s.size() == 2

    print("\n=== Test 4: Fill stack to capacity ===")
    s.push(40)
    s.push(50)
    s.push(60)
    print(s)
    assert s.is_full()
    try:
        s.push(70)
    except IndexError as e:
        print("Expected error:", e)

    print("\n=== Test 5: Pop all elements ===")
    while not s.is_empty():
        print("Popped:", s.pop())
    print(s)
    assert s.is_empty()

    print("\n=== Test 6: Initialize with values ===")
    s2 = ArrayStack(values=[1, 2, 3, 4])
    print(s2)
    assert s2.size() == 4
    assert s2.peek() == 4

    print("\nAll tests passed successfully ✅")


# Run the test
test_array_stack()
```

- **Linked List Stack**


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedListStack:
    def __init__(self, values=None):
        self.head = None
        self._size = 0
        if values:
            for val in values:
                self.push(val)

    def push(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
        self._size += 1

    def pop(self):
        if self.is_empty():
            raise IndexError("Stack underflow : cannot pop from empty stack")
        pop_data = self.head.data
        self.head = self.head.next
        self._size -= 1
        return pop_data

    def peek(self):
        if self.is_empty():
            raise IndexError("Stack underflow : cannot pop from empty stack")
        return self.head.data

    def is_empty(self):
        return self.head == None

    def size(self):
        return self._size

    def __repr__(self):
        elements = []
        current = self.head
        while current:
            elements += [current.data]
            current = current.next
        body = "  ↑ top \n" + "\n".join(f"| {elem} |" for elem in elements)
        return f"{body} \n(Size = {self._size})"


def test_linked_list_stack():
    print("=== TESTING LinkedListStack ===")

    # 1️⃣ Test empty stack
    s = LinkedListStack()
    assert s.is_empty() == True
    assert s.size() == 0
    print("✅ Empty stack test passed")

    # 2️⃣ Test push operation
    s.push(10)
    s.push(20)
    s.push(30)
    assert s.size() == 3
    assert s.peek() == 30  # top should be last pushed element
    print("✅ Push and peek test passed")

    # 3️⃣ Test pop operation
    popped = s.pop()
    assert popped == 30
    assert s.peek() == 20
    assert s.size() == 2
    print("✅ Pop test passed")

    # 4️⃣ Test stack initialized with values
    s2 = LinkedListStack(values=[1, 2, 3])
    assert s2.size() == 3
    assert s2.peek() == 3  # last element pushed
    print("✅ Initialization with values test passed")

    # 5️⃣ Test multiple pops until empty
    s2.pop()
    s2.pop()
    s2.pop()
    assert s2.is_empty() == True
    try:
        s2.pop()
    except IndexError:
        print("✅ Underflow (empty pop) test passed")

    # 6️⃣ Show representation
    print("\nStack representation:")
    print(s)

    print("\n🎉 All tests passed successfully!")


test_linked_list_stack()
```

```python
def symbol_check(symbol_str):
    symbol_stack = LinkedListStack()
    for symb in symbol_str:
        if symb in "([{":
            symbol_stack.push(symb)
        else:
            if symbol_stack.is_empty():
                return False
            else:
                last_symb = symbol_stack.pop()
                if not matches(last_symb, symb):
                    return False

    if not symbol_stack.is_empty():
        return False

    return True


def matches(opening, closing):
    opens = "([{"
    closes = ")]}"
    return opens.index(opening) == closes.index(closing)


symbol_check("[(])")
```

- **Array Queue** (Deck included)


```python
from array import array


class ArrayQueue:
    def __init__(self, values=None, dtype="i", capacity=0):
        if values and capacity == 0:
            capacity = len(values)

        self.capacity = capacity
        self.dtype = dtype
        self.queue = array(dtype, [0] * capacity)
        self.front = 0
        self.rear = -1
        self.size = 0
        if values:
            for val in values:
                self.enqueue_rear(val)

    def is_full(self):
        return self.size == self.capacity

    def is_empty(self):
        return self.size == 0

    def enqueue_rear(self, data):
        if self.is_full():
            raise IndexError("Queue overflow : cannot enqueue onto full queue")
        self.rear = (self.rear + 1) % self.capacity
        self.queue[self.rear] = data
        self.size += 1

    def enqueue_front(self, data):
        if self.is_full():
            raise IndexError("Queue overflow : cannot enqueue onto full queue")
        self.front = (self.front - 1 + self.capacity) % self.capacity
        self.queue[self.front] = data
        self.size += 1

    def dequeue_front(self):
        if self.is_empty():
            raise IndexError("Queue underflow : cannot dequeue from empty queue")
        value = self.queue[self.front]
        self.front = (self.front + 1) % self.capacity
        self.size -= 1
        return value

    def dequeue_rear(self):
        if self.is_empty():
            raise IndexError("Queue underflow : cannot dequeue from empty queue")
        value = self.queue[self.rear]
        self.rear = (self.rear - 1 + self.capacity) % self.capacity
        self.size -= 1
        return value

    def __repr__(self):
        if self.is_empty():
            return "[ empty queue ]"

        elements = []
        i = self.front
        for _ in range(self.size):
            elements.append(str(self.queue[i]))
            i = (i + 1) % self.capacity

        body = " <- ".join(elements)
        return f"front → {body} ← rear\n(capacity = {self.capacity}, dtype = '{self.dtype}')"


ArrayQueue([1, 2, 3, 4])
```

- **LinkedList Queue**


```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedListQueue:
    def __init__(self, values=None):
        self.front = None
        self.rear = None
        self.size = 0
        if values:
            for val in values:
                self.enqueue(val)

    def is_empty(self):
        return self.front is None

    def enqueue(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self.size += 1

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Queue underflow : cannot dequeue from empty queue")
        value = self.front.data
        self.front = self.front.next
        if not self.front:
            self.rear = None
        self.size -= 1
        return value

    def get_front(self):
        return self.front

    def get_rear(self):
        return self.rear

    def __len__(self):
        return self.size

    def __repr__(self):
        if self.is_empty():
            return "[ empty queue ]"
        elements = []
        current = self.front
        while current:
            elements += [str(current.data)]
            current = current.next

        body = " <- ".join(elements)
        return f"front -> {body} <- rear \n (size = {self.size})"


def test_linked_list_queue():
    print("=== TEST: LinkedListQueue ===")

    # 1️⃣ Create queue with initial values
    q = LinkedListQueue([1, 2, 3])
    print("Initial queue:", q)
    assert len(q) == 3, "Initial size should be 3"
    assert q.get_front().data == 1, "Front should be 1"
    assert q.get_rear().data == 3, "Rear should be 3"

    # 2️⃣ Enqueue a few elements
    q.enqueue(4)
    q.enqueue(5)
    print("After enqueue(4), enqueue(5):", q)
    assert len(q) == 5, "Size should be 5 after enqueues"
    assert q.get_rear().data == 5, "Rear should now be 5"

    # 3️⃣ Dequeue a few elements
    val1 = q.dequeue()
    val2 = q.dequeue()
    print(f"Dequeued {val1} and {val2}")
    print("Queue after dequeues:", q)
    assert val1 == 1 and val2 == 2, "Dequeued elements should be 1, then 2"
    assert q.get_front().data == 3, "Front should now be 3"
    assert len(q) == 3, "Size should be 3 after two dequeues"

    # 4️⃣ Enqueue again after dequeues
    q.enqueue(6)
    print("After enqueue(6):", q)
    assert q.get_rear().data == 6, "Rear should now be 6"
    assert len(q) == 4, "Size should be 4 after enqueue"

    # 5️⃣ Dequeue all elements
    while not q.is_empty():
        print("Dequeued:", q.dequeue())
    print("Queue after emptying:", q)
    assert q.is_empty(), "Queue should be empty now"

    # 6️⃣ Try to dequeue from empty queue — should raise error
    try:
        q.dequeue()
        raise AssertionError("Expected IndexError on dequeue from empty queue")
    except IndexError:
        print("✅ Correctly raised IndexError on empty dequeue")

    print("=== ALL TESTS PASSED ===")


test_linked_list_queue()
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class LinkedListDeQueue:
    def __init__(self, values=None):
        self.front = None
        self.rear = None
        self.size = 0
        if values:
            for val in values:
                self.enqueue_rear(val)

    def is_empty(self):
        return self.front is None

    def enqueue_rear(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            new_node.prev = self.rear
            self.rear = new_node
        self.size += 1

    def enqueue_front(self, data):
        new_node = Node(data)
        if self.is_empty():
            self.front = self.rear = new_node
        else:
            self.front.prev = new_node
            new_node.next = self.front
            self.front = new_node
        self.size += 1

    def dequeue_front(self):
        if self.is_empty():
            raise IndexError("Queue underflow : cannot dequeue from empty queue")
        value = self.front.data
        self.front = self.front.next
        if self.front:
            self.front.prev = None
        else:
            self.rear = None
        self.size -= 1
        return value

    def dequeue_rear(self):
        if self.is_empty():
            raise IndexError("Queue underflow : cannot dequeue from empty queue")
        value = self.rear.data
        self.rear = self.rear.prev
        if self.rear:
            self.rear.next = None
        else:
            self.front = None
        self.size -= 1
        return value

    def get_front(self):
        return self.front

    def get_rear(self):
        return self.rear

    def __len__(self):
        return self.size

    def __repr__(self):
        if self.is_empty():
            return "[ empty queue ]"
        elements = []
        current = self.front
        while current:
            elements += [str(current.data)]
            current = current.next

        body = " <- ".join(elements)
        return f"front -> {body} <- rear \n (size = {self.size})"
```

- **Complete B-Tree** (rough approach)


```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class Tree:
    def __init__(self, values=None):
        self.root = None
        self.nodes = []
        if values:
            for val in values:
                self.nodes += [TreeNode(val)]
            self.root = self.nodes[0]
            self.build_tree()

    def build_tree(self):
        n = len(self.nodes)
        for i in range(n):
            left = 2 * i + 1
            right = 2 * i + 2
            if left < n:
                self.nodes[i].left = self.nodes[left]
            if right < n:
                self.nodes[i].right = self.nodes[right]

    def __iter__(self):
        yield from self.traverse(self.root)

    def traverse(self, node):
        if node:
            yield node.data
            yield from self.traverse(node.left)
            yield from self.traverse(node.right)


tree = Tree([1, 2, 3, 4, 5, 6, 7])
for i in tree:
    print(i)
# print(tree.root.right.data)
```

- **Complete B-Tree**


```python
from collections import deque


class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class CompleteBTree:
    def __init__(self, values=None):
        self.root = None
        if values:
            for val in values:
                self.insert(val)

    def insert(self, data):
        new_node = TreeNode(data)
        if not self.root:
            self.root = new_node
            return

        deck = deque([self.root])
        while deck:
            current = deck.popleft()

            if not current.left:
                current.left = new_node
                return
            else:
                deck.append(current.left)

            if not current.right:
                current.right = new_node
                return
            else:
                deck.append(current.right)

    def pre_order_traversal(self):
        node = self.root
        stack = []
        elements = []
        while True:
            while node:
                elements.append(node.data)
                stack.append(node)
                node = node.left
            if not stack:
                break
            node = stack.pop()
            node = node.right

        return elements

    def inorder_traversal(self):
        node = self.root
        stack = []
        elements = []
        while True:
            while node:
                stack.append(node)
                node = node.left
            if not stack:
                break
            node = stack.pop()
            elements.append(node.data)
            node = node.right

        return elements

    def post_order_traversal(self):
        node = self.root
        stack = []
        elements = []
        prev = None
        while stack or node:
            while node:
                stack.append(node)
                node = node.left

            node = stack[-1]

            if not node.right or node.right == prev:
                elements.append(node.data)
                stack.pop()
                prev = node
                node = None
            else:
                node = node.right

        return elements

    def level_order_traversal(self):
        if not self.root:
            return []

        deck = deque([self.root])
        elements = []

        while deck:
            node = deck.popleft()
            elements.append(node.data)
            if node.left:
                deck.append(node.left)
            if node.right:
                deck.append(node.right)

        return elements

    def max_depth(self):
        def compute_depth(node):
            if not node:
                return 0
            return max(compute_depth(node.left), compute_depth(node.right)) + 1

        return compute_depth(self.root)

    def max_depth(self):
        if not self.root:
            return "< empty tree >"

        deck = deque([(self.root, 1)])
        maxDepth = 0

        while deck:
            node, depth = deck.popleft()
            if depth > maxDepth:
                maxDepth = depth
                data = node.data

            if node.left:
                deck.append((node.left, depth + 1))
            if node.right:
                deck.append((node.right, depth + 1))

        return maxDepth, data

    def remove(self, value):
        if not self.root:
            raise IndexError("cannot delete from empty tree")

        if self.root.data == value:
            self.root = None
            return

        deck = deque([self.root])

        while deck:
            node = deck.popleft()
            if node.left:
                if node.left.data == value:
                    node.left = None
                    return
                else:
                    deck.append(node.left)

            if node.right:
                if node.right.data == value:
                    node.right = None
                    return
                else:
                    deck.append(node.right)

        raise ValueError(f" value : {value} not found in tree")


tree = CompleteBTree([1, 45, 3, 9, 5, 3, 4, 5, 6, 7, 9])
print(tree.pre_order_traversal())
print(tree.inorder_traversal())
print(tree.post_order_traversal())
print(tree.level_order_traversal())
tree.remove(3)
print(tree.level_order_traversal())
print(tree.max_depth())
```

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BST:
    def __init__(self, values=None):
        self.root = None
        if values:

            def midorder(arr):
                if not arr:
                    return []

                mid = len(arr) // 2
                return [arr[mid]] + midorder(arr[:mid]) + midorder(arr[mid + 1 :])

            values = midorder(sorted(values))
            self.root = TreeNode(values.pop(0))
            for val in values:
                self.insert(val)

    def insert(self, value):
        new_node = TreeNode(value)

        if not self.root:
            self.root = new_node
            return

        current = self.root
        parent = None
        while current:
            parent = current
            if value < current.data:
                current = current.left
            elif value > current.data:
                current = current.right
            else:
                print(f"value : {value} already present")
                return
        if value < parent.data:
            parent.left = new_node
        else:
            parent.right = new_node

    def inorder_traversal(self):
        if not self.root:
            return "< empty stack >"
        stack = []
        elements = []
        current = self.root

        while True:
            while current:
                stack.append(current)
                current = current.left

            if not stack:
                break

            current = stack.pop()
            elements.append(current.data)
            current = current.right

        return elements

    # def inorder_traversal(self):
    #     elements = []
    #     def _inorder(node):
    #         if not node:
    #             return
    #         _inorder(node.left)
    #         elements.append(node.data)
    #         _inorder(node.right)
    #     _inorder(self.root)
    #     return elements

    def cum_sum_tree(self):
        if not self.root:
            raise ValueError("cant perform cum_sum_tree() in empty tree")
        stack = []
        current = self.root
        total = 0

        while True:
            while current:
                stack.append(current)
                current = current.right

            if not stack:
                break

            current = stack.pop()
            total += current.data
            current.data = total
            current = current.left


bst = BST([5, 7, 10, 15, 20])
print(bst.inorder_traversal())
bst.cum_sum_tree()
print(bst.inorder_traversal())
```

```python
def bubble_sort(arr):
    n = len(arr)
    count = 0
    for i in range(n):
        print(f"iteration : {count}")
        swapped = False
        for j in range(n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
        count += 1
        print(arr)


arr = [1, 2, 3, 4, 5, 5, 2, 5, 3, 6, 7, 6]
bubble_sort(arr)
arr
```

# Hash Table


```python
class HashTable:
    def __init__(self, size):
        self.size = size
        self.table = [[] for _ in range(size)]

    def __hash_function(self, key):
        return sum(ord(ch) for ch in key) % self.size

    def insert(self, key, value):
        idx = self.__hash_function(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                pair[1] = value
                return
        self.table[idx].append([key, value])

    def search(self, key):
        idx = self.__hash_function(key)
        for pair in self.table[idx]:
            if pair[0] == key:
                return pair[1]

        raise ValueError(f"key : {key} not fouund")

    def __repr__(self):
        lines = []
        for i, chain in enumerate(self.table):
            lines.append(f"Index {i}: {chain}")
        return "\n".join(lines)


def test_hash_table():
    # Create a hash table of size 5
    ht = HashTable(5)
    print("=== Initial empty table ===")
    print(ht)
    print()

    # Insert key-value pairs
    ht.insert("apple", 10)
    ht.insert("banana", 20)
    ht.insert("orange", 30)
    ht.insert("grape", 40)
    ht.insert("melon", 50)

    print("=== After inserting 5 items ===")
    print(ht)
    print()

    # Test search
    print("Search for 'banana':", ht.search("banana"))
    print("Search for 'orange':", ht.search("orange"))
    print()

    # Update a value
    ht.insert("banana", 99)
    print("=== After updating 'banana' ===")
    print(ht)
    print("Search for updated 'banana':", ht.search("banana"))
    print()

    # Test collision handling (keys that hash to the same index)
    ht.insert("papel", 123)  # an anagram of "apple" → likely same hash
    print("=== After inserting 'papel' (possible collision) ===")
    print(ht)
    print()

    # Test missing key (expect error)
    try:
        ht.search("cherry")
    except ValueError as e:
        print("Error caught:", e)


test_hash_table()
```

# Heap


```python
class MaxHeap:
    def __init__(self) -> None:
        self.data = []
        self._len = 0

    def push(self, el):
        self.data.append(el)
        self._len += 1
        i = len(self.data) - 1
        while i > 0:
            parent = (i - 1) // 2
            if self.data[parent] >= self.data[i]:
                break
            self.data[parent], self.data[i] = self.data[i], self.data[parent]
            i = parent

    def __len__(self):
        return self._len

    def pop(self):
        top = self.data[0]
        last = self.data.pop()
        if self.data:
            self.data[0] = last
            self._sift_down()
        return top

    def _sift_down(self):
        n = len(self.data)
        i = 0

        while True:
            largest = i
            left, right = 2 * i + 1, 2 * i + 2

            if left < n and self.data[left] > self.data[largest]:
                largest = left

            if right < n and self.data[right] > self.data[largest]:
                largest = right

            if largest == i:
                return

            self.data[i], self.data[largest] = self.data[largest], self.data[i]
            i = largest


heap = MaxHeap()
for value in [5, 3, 8, 1, 9, 2]:
    heap.push(value)

print("Heap array:     ", heap.data)
for _ in range(6):
    pop_val = heap.pop()
    print(f"Popped value: {pop_val}")
    print(f"Heap after pop: {heap.data}")
```

# LRU Cache


```python
class Node:
    def __init__(self, key=0, value=0):
        self.key = key
        self.value = value
        self.prev = None
        self.next = None

class LRUCache:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = {}  # hash map - key to Node
        
        # Dummy head and tail nodes to avoid edge cases during insertion/deletion
        # explanation below
        self.head = Node()
        self.tail = Node()
        self.head.next = self.tail
        self.tail.prev = self.head

    def _remove(self, node: Node):
        prev_node = node.prev
        next_node = node.next
        prev_node.next = next_node
        next_node.prev = prev_node

    def _insert(self, node: Node):
        node.prev = self.head
        node.next = self.head.next
        self.head.next.prev = node
        self.head.next = node

    def get(self, key: int) -> int:
        if key in self.cache:
            node = self.cache[key]
            # mark as recently used by moving it to the head
            self._remove(node)
            self._insert(node)
            return node.value
        return -1

    def put(self, key: int, value: int) -> None:
        if key in self.cache:
            # remove old node if updating an existing key
            self._remove(self.cache[key])
        
        # create and insert node
        new_node = Node(key, value)
        self.cache[key] = new_node
        self._insert(new_node)
        
        # evict LRU if more than capacity
        if len(self.cache) > self.capacity:
            lru = self.tail.prev
            self._remove(lru)
		 # Node.key helps to fetch it in O(1)
            del self.cache[lru.key]

    def __repr__(self):
        elements = []
        current = self.head.next

        while current != self.tail:
            elements.append(f"{current.key}: {current.value}")
            current = current.next

        return "LRUCache([" + " <-> ".join(elements) + "])"

```

```python
cache = LRUCache(4)

cache.put(1, 100)
cache.put(2, 200)
cache.put(3, 300)
cache.put(4, 400)

print(cache)

print(cache.get(2))  # 200

print(cache)

cache.put(5, 500)

print(cache)

print(cache.get(1))  
print(cache.get(3))  

print(cache)

cache.put(2, 250)    

print(cache)

print(cache.get(2))  # 250
print(cache.get(4))  # 400

print(cache)
```

```python
def can_attend_all(events):
   
    # sort events by their start time
    sorted_events = sorted(events, key=lambda x: x[0])

    # check overlapping intervals
    for i in range(1, len(sorted_events)):
        # if current event starts before the previous event ends => overlap
        if sorted_events[i][0] < sorted_events[i - 1][1]:
            return False

    return True


def min_rooms_required(events):

    if not events:
        return 0

    time_points = []
    for start, end in events:
        time_points.append((start, 1))  # add one room
        time_points.append((end, -1))  # subtract one room

    # sorted by time, if times are equal sort by type (-1 before 1).
    # room freed before a new event at the overlapping time
    time_points.sort(key=lambda x: (x[0], x[1]))

    curr_rooms = 0
    max_rooms = 0

    for _, room in time_points:
        curr_rooms += room
        max_rooms = max(curr_rooms, max_rooms)

    return max_rooms
```

```python
events = [(9, 10), (10, 11), (11, 12)]

print("Can attend all events:", can_attend_all(events))
print("Minimum rooms required:", min_rooms_required(events))
```

```python
import heapq


def min_rooms_required_heap(events):
    if not events:
        return 0

    sorted_events = sorted(events, key=lambda x: x[0])

    # min-heap keep track of meeting end times
    free_rooms = []

    # first meeting end time 
    heapq.heappush(free_rooms, sorted_events[0][1])

    for event in sorted_events[1:]:
        start, end = event

        # if the earliest ending meeting finishes before or at the exact time
        # the current meeting starts => reuse  room.
        if free_rooms[0] <= start:
            heapq.heappop(free_rooms)

        # allocate a room for the current meeting (this either adds a new room
        # or takes over the one we just popped)
        heapq.heappush(free_rooms, end)

    return len(free_rooms)
```
