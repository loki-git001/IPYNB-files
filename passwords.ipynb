```python
import random
import string
import os
import time

# Character set: a-z, A-Z, 0-9
chars = string.ascii_letters + string.digits  

# Configuration
num_files = 10
lines_per_file = 10_000_000
output_dir = "combinations_files"

# Make sure the output directory exists
os.makedirs(output_dir, exist_ok=True)

for file_num in range(1, num_files + 1):
    output_file = os.path.join(output_dir, f"combinations_part_{file_num}.txt")
    start_time = time.time()  # start timer for this file
    with open(output_file, "w", encoding="utf-8") as f:
        for i in range(lines_per_file):
            combo = "".join(random.choices(chars, k=10))
            f.write(combo + "\n")

            # Progress feedback every 1 million lines
            if (i + 1) % 1_000_000 == 0:
                elapsed = time.time() - start_time
                avg_per_million = elapsed / ((i + 1) / 1_000_000)
                remaining_million = (lines_per_file - (i + 1)) / 1_000_000
                est_remaining = remaining_million * avg_per_million
                print(f"File {file_num}: {i + 1:,} lines written "
                      f"(Elapsed: {elapsed:.1f}s, "
                      f"Est remaining: {est_remaining:.1f}s)")

    total_elapsed = time.time() - start_time
    print(f"✅ Finished writing {output_file} in {total_elapsed:.1f}s")

print(f"All {num_files} plain text files written in '{output_dir}'")
```
