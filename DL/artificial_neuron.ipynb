# Artificial Neurons

A neuron (artificial neuron) is:

👉 A basic computational unit in a neural network that receives inputs, processes them using parameters, and produces an output through an activation function.


```python
import math


def sigmoid(x):
    """Sigmoid activation function"""
    return 1.0 / (1 + math.exp(-x))


def activate(inputs, weights):
    """
    Simulates the computational unit of the artificial neuron.
    Takes in inputs and weights, calculates the net input,
    and applies the activation function.
    """
    # 1. Perform Net Input
    h = 0
    for x, w in zip(inputs, weights):
        h += x * w

    # 2. Perform Activation
    return sigmoid(h)


if __name__ == "__main__":
    # Define sample inputs and corresponding weights
    inputs = [0.5, 0.3, 0.2]
    weights = [0.4, 0.7, 0.2]

    # Calculate neuron output
    output = activate(inputs, weights)

    print(f"Neuron Output: {output}")
```

# Implementing an Artificial Neuron From Scratch

## 1. Biological vs. Artificial Neurons

- **Biological Neurons:** Complex systems primarily composed of **dendrites** (input modules that receive signals), a **cell body** (operational center that modulates the signal), and an **axon** (transmits the signal to other neurons).
- **Artificial Neurons:** A simplified mathematical model inspired by biological neurons. They take a series of **inputs** (equivalent to dendrites), apply **weights**, and pass them through a processing unit (equivalent to the cell body) to produce an **output**.

## 2. The Math Behind the Neuron

An artificial neuron processes data in two main phases:

### Phase 1: The Net Input (Linear Combination)

The neuron calculates the sum of all inputs multiplied by their respective weights.

$$h = \sum_{i} (x_i \cdot w_i)$$

_(Where $x$ represents the inputs, $w$ represents the weights, and $h$ is the net input)_

### Phase 2: The Activation Function

The net input is passed through an activation function to modulate the final output. A very common activation function is the **Sigmoid function**, which smoothly bounds the output between 0 and 1, making it easy to differentiate.

$$y = \frac{1}{1 + e^{-h}}$$
