## Euclidean Distance

The Euclidean distance between two points is given by:

$$
d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}
$$

---

## Manhattan Distance

The Manhattan distance between two points is given by:

$$
d = |x_1 - x_2| + |y_1 - y_2|
$$

```python
import numpy as np

# Euclidean Distance

dist = 0 

p1 = (1,1,1,1)
p2 = (2,2,2,2)

for i in range(len(p1)):
    dist += (p1[i] - p2[i])**2

np.sqrt(dist)
```

```python
# Manhattan distance

dist = 0 

p1 = (1, 1, 1, 1)
p2 = (2, 2, 2, 2)

for i in range(len(p1)):
    dist += abs(p1[i] - p2[i])

dist
```
