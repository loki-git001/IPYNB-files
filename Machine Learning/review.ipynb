```python
from typing import Any


class NN(nn.Module):
    def __init__(self, *args: Any, **kwargs: Any) -> None:
        super().__init__(*args, **kwargs)

        self.sequential = nn.Sequential(
            nn.LazyLinear(10),
            nn.ReLU(),
            nn.LazyLinear(7),
            nn.ReLU(),
            nn.LazyLinear(3),
            nn.ReLU(),
            nn.LazyLinear(1)
        )
    def forward(self, x):
        return self.sequential(x)
```
