#### General Mathematical Form of the Signal

The signals in the code follow the standard sinusoidal form:

$$
x(t) = A \cdot \sin\left(2\pi (f t - \phi)\right)
$$

---

### Meaning of Variables

* $x(t)$ : Signal amplitude at time $t$
* $A$ : Amplitude (controls the height of the wave)
* $f$ : Frequency in Hertz (cycles per second)
* $t$ : Time in seconds
* $\phi$ : Phase shift (controls horizontal shift of the wave)

---

### Additional Notes

* The term $2\pi f$ represents the **angular frequency**
* The expression $(f t - \phi)$ determines how the wave evolves over time
* Changing $\phi$ shifts the signal left or right without altering its shape

---


```python
# create a sinusoid

f = 523
phase = 0
phase2 = 0.2

sin = 0.5 * np.sin(2 * np.pi * (f * t - phase))
sin2 = 0.5 * np.sin(2 * np.pi * (f * t - phase2))

plt.figure(figsize=(18, 8))
plt.plot(t[10000:10400], sin[10000:10400], color="r")
plt.plot(t[10000:10400], sin2[10000:10400], color="y")


plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.show()
```

```python
f = 256
phase = 0.9

sin = 0.1 * np.sin(2*np.pi * (f*t - phase)) 

plt.figure(figsize=(18, 8))
plt.plot(t[10000:10400], signal[10000:10400])
plt.plot(t[10000:10400], sin[10000:10400], color='r')
plt.fill_between(t[10000:10400], sin[10000:10400] * signal[10000:10400], color="y")

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.show()
```
