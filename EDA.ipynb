```python
df["quality"].value_counts()
sns.barplot(df["quality"].value_counts())
plt.xlabel("Quality")
plt.ylabel("Count")
plt.show()
```

```python
sns.set_style("darkgrid")

num_columns = df.select_dtypes(include = ["int64", "float64"])

plt.figure(figsize=(14, len(num_columns)*3))
for idx, col in enumerate(num_columns, 1):
    plt.subplot(len(num_columns), 2, idx)
    sns.histplot(df[col], kde = True)
    plt.title(f"{col} | Skewness: {round(df[col].skew(), 2)}")

plt.tight_layout()
plt.show()
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue = "citric acid")
```

```python
sns.violinplot(data=df, x="quality", y="alcohol", palette="Set1", hue="quality", legend=False)

plt.title("Violin plot of quality and alcohol")
plt.xlabel("Quality")
plt.ylabel("Alcohol")
plt.show()
```

```python
sns.violinplot([10, 12, 11, 9, 400, 450, 500, 470, 480, 490])
```

```python
plt.figure(figsize=(10, 10))
sns.heatmap(df.corr(), annot=True, fmt='.2f', linewidths=2)
```

The below graph shows how the average value of each feature changes across different wine-quality ratings.

```python
df.groupby("quality").mean().transpose().plot(figsize=(10,15))
plt.show()
df.groupby("quality").mean()
```
