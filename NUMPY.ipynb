```python
# Create a large 1-D array x of 1,000,000 random floats in [0,1) using default_rng(0).random(1_000_000). 
# Compute the sum in three ways and time them: Python loop, np.sum, and x @ np.ones_like(x). 
# Print the times and confirm all sums match within atol=1e-93
```

```python
import time
arr = rng.random(1000000)
sum1 = 0
#---------Python Loop----------
t1_start = time.time()
sum1 = 0
for i in arr:
    sum1 += i
t1_end = time.time()


#---------np.sum----------------
t2_start = time.time()
sum2 = np.sum(arr)
t2_end = time.time()


#--------Dot product------------
t3_start = time.time()
sum3 = arr @ np.ones_like(arr)
t3_end = time.time()


print("--------PYTHON LOOP--------")
print(f"Sum : {sum1} ")
print(f"Start time : {t1_start} \n End time : {t1_end}")

print("--------NP SUM--------")
print(f"Sum : {sum2} ")
print(f"Start time : {t2_start} \n End time : {t2_end}")

print("--------DOT PRODUCT--------")
print(f"Sum : {sum3} ")
print(f"Start time : {t3_start} \n End time : {t3_end}")


print(f"\n\n\n {np.allclose(sum1, sum2, sum3, atol = 1e-93)}")
```

```python
# Given a 2-D array a with shape (4,5), compute the mean of each row, then subtract the corresponding row mean from each row (mean-centering). 
# Verify each resulting row has mean≈0 within atol=1e-12. Print the centered array and the verification result.
# Targets: broadcasting, keepdims/shape alignment, np.mean(axis=1), np.allclose
```

```python
arr = np.arange(1,21).reshape(4,5)
mean = arr.mean(axis = 1, keepdims = True)
mean_center = arr - mean
verify = np.allclose(mean_center.mean(axis=1), 0, atol = 1e-12)
print(mean_center)
print(verify)
```

```python
# Create two arrays: a = np.array([1, 2, 0, -1, -2]) and b = np.array([0, 1, 2, 0, -2]),
# Compute the elementwise safe division a / b where division by zero returns 0 instead of inf/NaN, without using Python loops. 
# Print the result and its dtype.
# Targets: ufuncs, np.divide with where and out parameters
```

```python
a = np.array([1, 2, 0, -1, -2])
b = np.array([0, 1, 2, 0, -2])
result = np.zeros_like(a, dtype=float)
np.divide(a, b, out=result, where=b!=0)
print(result)
print(result.dtype)
```

```python
# Given an array x = np.array([0.0, 1.0, 2.0, np.nan, 4.0, np.inf, -np.inf]), 
# compute the elementwise logarithm base e with safe handling:

#     For x<=0 or non-finite values, return 0.

#     Otherwise return np.log(x).

# Do this without Python loops, using ufuncs and masking. Print the result and verify no NaN/inf remains.
```

```python
x = np.array([0.0, 1.0, 2.0, np.nan, 4.0, np.inf, -np.inf])
y = np.nan_to_num(x, nan=0, posinf=0, neginf=0)
mask = (x > 0) & np.isfinite(x)
y[mask] = np.log(x[mask])
y
```

```python
# Create a 6×6 NumPy array with values from 0 to 35.
# Using fancy indexing (no Python loops), set all elements in a 3×3 central block to 99.
# Print the modified array.
# Verify that only those 9 positions were changed, and all other elements remain unchanged.
# Targets: fancy indexing with row/column index arrays, broadcasting behavior of indexing
```

```python
matrix = np.arange(0,36).reshape(6, 6)
rows = [1, 3, 5]
cols = [0, 2, 4]
matrix[np.ix_(rows, cols)] = 99
print(matrix)

#----------verify------------
check = np.ones_like(matrix, dtype=bool)
check[np.ix_(rows, cols)] = False
unchanged = np.all(matrix[check] != 99)
changed = np.all(matrix[~check] == 99)
print(changed & unchanged)
```

```python
# Given a 10,000-length 1-D float array x, compute a rolling mean with window size 5 using only NumPy (no loops, no pandas). 
# Return an array of length 9,996 containing the valid-centered means. 
# Print the result’s first 5 values and its shape.
```

```python
rng = np.random.default_rng()
x = rng.random(10000)
w_size = 5
stride = np.lib.stride_tricks.sliding_window_view(x, w_size)
rolling_mean = stride.mean(axis = 1)
print(rolling_mean[:5])
print(rolling_mean.shape)
```
