```python
def fact(n):
    if n == 0:
        return 0
    elif n == 1 :
        return 1
    else :
        return n * (fact(n-1))


fact(5)
```

Is Palindrome

```python
def is_palindrome(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return is_palindrome(s[1:-1])

is_palindrome("ababsg")
```

- **Singly linked list**

```python
from collections.abc import Iterable

class LinkedList:
    def __init__(self, values = None):
        self.head = None
        if values:
            for val in values:
                self.append(val)

# Append a single element to array  
    def append(self, value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = new_node

# Append an iterable
    def extend(self, iterable):
        if not isinstance(iterable, Iterable):
            raise TypeError("extend() method only works with an iterable")
        for val in iterable:
            self.append(val)
            
#Join two linked lists
    def join(self, ll):
        if not isinstance(ll, LinkedList):
            raise ValueError("join() expects another linked list")
        if not self.head:
            self.head = ll.head
            return
        if not ll.head:
            return
        current = self.head
        while current.next:
            current = current.next
        current.next = ll.head
    
# Insert an element in between
    def insert(self, index, value):
        new_node = Node(value)

        if index == 0:
            new_node.next = self.head
            self.head = new_node
            return

        count = 0
        current = self.head

        while count < index - 1 :
            if not current:
                raise IndexError("Linked list index out of bounds")
            current = current.next
            count +=1 


        new_node.next = current.next
        current.next = new_node

# Insert an iterable
    def insert_many(self, index, iterable):
        if not isinstance(iterable, Iterable):
            raise ValueError("insert_many() expects an iterable value")

        nodes = [Node(val) for val in iterable]
        for i in range(len(nodes) - 1):
            nodes[i].next = nodes[i+1]

        if index == 0:
            nodes[-1].next = self.head
            self.head = nodes[0]
            return

        count = 0
        current = self.head

        while count < index - 1:
            if not current:
                raise IndexError("Linked List index out of bounds")
            current = current.next
            count += 1

        nodes[-1].next = current.next
        current.next = nodes[0]

# Delete an element by index
    def delete(self, index):
        if not self.head:
            raise IndexError("Can't delete from empty list")
        
        if index == 0:
            self.head = self.head.next
            return

        current = self.head
        count = 0
        while count < index - 1:
            if not current:
                raise IndexError("Linked list index out of bounds")
            current = current.next
            count += 1 
                
        current.next = current.next.next 

#Delete first occurence of an element by value
    def remove(self, value):
        current = self.head
        previous = None
        while current:
            if current.data == value:
                if not previous:
                    self.head = current.next
                else:
                    previous.next = current.next
                return

            previous = current
            current = current.next
        raise ValueError(f"{value} not found in the Linked List")

# Sort the linked list
    def sort(self, reverse = False):
        values = list(self)
        values.sort(reverse = reverse)
        self.head = None
        for val in values:
            self.append(val)

# Reverse the linked list
    def reverse(self):
        prev = None
        current = self.head
        while current:
            next_node = current.next
            current.next = prev
            prev = current
            current = next_node
        
        self.head = prev   

# Remove duplicates
    def remove_dups(self):
        seen = set()
        current = self.head
        prev = None
        while current:
            if current.data in seen:
                prev.next = current.next
            else:
                seen.add(current.data)

            prev = current
            current = current.next
            

    
    def __getitem__(self, index):
        count = 0
        current = self.head
        while current:
            if count == index:
                return current.data
            count += 1
            current = current.next
        raise IndexError("Linked list index out of bounds")

    def __setitem__(self, index, value):
        count = 0
        current = self.head
        while current:
            if count == index:
                current.data = value
                return
            current = current.next
            count += 1
        raise IndexError("Linked list index out of bounds")

    def __len__(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    def __iter__(self):
        current = self.head
        while current:
            yield current.data
            current = current.next

    def __repr__(self):
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        return "[" + " -> ".join(elements) + "]"


```

- **Doubly linked list**

```python
from collections.abc import Iterable

class DoublyLinkedList:
    def __init__(self, values = None):
        self.head = None
        self.tail = None
        if values:
            for val in values:
                self.append(val)

# Append a single value
    def append(self, value):
        new_node = Node(value)
        if not self.head:
            self.head = self.tail = new_node
            return
        self.tail.next = new_node
        new_node.prev = self.tail
        self.tail = new_node

# Append an iterable
    def extend(self, iterable):
        if not isinstance(iterable, Iterable):
            raise ValueError("extend() method only works with an iterable")
        for val in iterable:
            self.append(val)

# Insert an element at given index
    def insert(self, index, data):
        new_node = Node(data)
        
        if index == 0:
            new_node.next = self.head
            if self.head:
                self.head.prev = new_node
            else:
                self.tail = new_node
            self.head = new_node
            return

        current = self.head
        count = 0

        while count < index - 1:
            if not current:
                raise IndexError("Linked list index out of bounds")
            current = current.next
            count += 1

        new_node.next = current.next
        new_node.prev = current

        if current.next:
            current.next.prev = new_node
        else:
            self.tail = new_node
        current.next = new_node

# Insert an iterable at given index
    def insert_many(self, index, iterable):
        if not isinstance(iterable, Iterable):
            raise ValueError("insert_many() expects an iterable input")

        nodes = [Node(val) for val in iterable]
        for i in range(len(nodes)-1):
            nodes[i].next, nodes[i+1].prev = nodes[i+1], nodes[i]

        if index == 0:
            nodes[-1].next = self.head
            if self.head:
                self.head.prev = nodes[-1]
            else:
                self.tail = nodes[-1]
            self.head = nodes[0]
            return

        current = self.head
        count = 0
        
        while count < index-1:
            if not current:
                raise IndexError("Index of the linked list out of bounds")
            current = current.next
            count += 1

        nodes[-1].next = current.next
        nodes[0].prev = current

        if current.next:
            current.next.prev = nodes[-1]
        else:
            self.tail = nodes[-1]
        current.next = nodes[0]
        
# Delete an item by index
    def delete(self, index):
        if not self.head:
            raise IndexError("Cannot delete from empty list")
        
        current = self.head
        count = 0
        while count < index :
            if not current:
                raise IndexError("Linked List index out of bounds")
            current = current.next
            count += 1

        if current.prev:
            current.prev.next = current.next
        else:
            self.head = current.next

        if current.next:
            current.next.prev = current.prev
        else:
            self.tail = current.prev
        
# Delete the first occurence of an element
    def remove(self, value):
        if not self.head:
            raise IndexError("Cannot delete from empty list")

        current = self.head
        while current:
            if current.data == value:
                if current.prev:
                    current.prev.next = current.next
                else:
                    self.head = current.next

                if current.next:
                    current.next.prev = current.prev
                else :
                    self.tail = current.prev
                return
            current = current.next
        raise ValueError(f"{value} not found in the list")


# Sort the linked list
    def sort(self, reverse=False):
        values =[]
        current = self.head
        while current:
            values.append(current.data)
            current = current.next
        
        values.sort(reverse = reverse)
        self.head = self.tail = None
        for val in values:
            self.append(val)


# Reverse the linked list
    def reverse(self):
        current = self.head
        while current:
            current.next, current.prev = current.prev, current.next
            if not current.prev:
                self.tail = self.head
                self.head = current
            current = current.prev

    def __getitem__(self, index):
        current = self.head
        count = 0
        while current:
            if count == index:
                return current.data
            current = current.next
            count += 1
        raise IndexError("Linked List Index out of bounds")


    def __setitem__(self, index, value):
        current = self.head
        count = 0
        while current:
            if count == index:
                current.data = value
                return
            current = current.next
            count += 1
        raise IndexError("Index out of bounds")


    def __len__(self):
        count = 0 
        current = self.head
        while current:
            current = current.next
            count += 1
        return count


    def __iter__(self):
        current = self.head
        while current:
            yield current.data
            current = current.next

        
    def __repr__(self):
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next

        return "[" + " <-> ".join(elements) + "]"
```

- **Circular Linked List**

```python
from collections.abc import Iterable

class CircularLinkedList:
    def __init__(self, values=None):
        self.head = None
        if values:
            for val in values:
                self.append(val)

# Append a single node
    def append(self, value):
        new_node = Node(value)
        if not self.head:
            self.head = new_node
            new_node.next =  self.head
            return

        current = self.head
        while current.next != self.head:
            current = current.next

        current.next = new_node
        new_node.next = self.head

# Append multiple elements (iterable)
    def extend(self, iterable):
        if not isinstance(iterable, Iterable):
            raise ValueError("extend() expects an iterable as input")
        for val in iterable:
            self.append(val)

# Insert at a specific index
    def insert(self, index, value):
        new_node = Node(value)

        current = self.head
        count = 0
        while count < index-1:
            if current.next == self.head:
                raise IndexError("Index out of bounds")
            current = current.next
            count += 1

        current.next, new_node.next = new_node, current.next

    def __iter__(self):
        current = self.head
        while True:
            yield current.data
            current = current.next
            if current == self.head:
                break

            

```

Task: Find the Smallest Missing Positive Number in a Sorted Array

Problem Statement:

You are given a sorted list of unique positive integers. Some numbers may be missing in the sequence. Your task is to find the smallest missing positive number using binary search, not linear search.

Example:

Input: [1, 2, 3, 5, 6, 7]
Output: 4

Input: [2, 3, 4, 7, 8]
Output: 1

Requirements:

- Use binary search to solve the problem efficiently.
- The solution should run in O(log n) time.
- Avoid using any loops that scan the array linearly.

```python
def missing_seq(arr):

    left, right = 0, len(arr)-1

    while left <= right:
        mid = (left + right) //2
        if arr[mid] == mid + 1:
            left = mid+1
        else:
            right = mid-1
            
    return left+1

missing_seq([1,4,5,6,8])
```

**Longest Increasing Subarray** – contiguous elements that are strictly increasing

Example: [1, 2, 2, 3, 4] → [2, 3, 4]

```python
def longest_seq_subarray(arr):
    start = 0
    max_len = 1
    max_start = 0
    for i in range(1, len(arr)):
        if arr[i] <= arr[i-1]:
            start = i

        curr_len = i - start + i
        if max_len < curr_len:
            max_len = curr_len
            max_start = start

    return arr[max_start : max_start + max_len]

longest_seq_subarray([1, 2, 2, 3, 4])
```

1. You are given a sorted array nums of integers and a target value.
Write a function to return the index of the target if it exists in the array.
If it does not exist, return -1.

Input: nums = [1, 3, 5, 7, 9], target = 5  
Output: 2


```python
def target_index(arr, target):
    left, right = 0, len(arr)-1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
            right = mid-1
        else:
            left = mid + 1
    return -1

target_index([1, 5, 6, 7, 9], 5)
```

2. Write a function to reverse a singly linked list.

Input: 1 -> 2 -> 3 -> None  
Output: 3 -> 2 -> 1 -> None


```python
def reverse_ll(ll):
    current = ll.head
    prev = None
    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node
    ll.head = prev

    return ll

ll = LinkedList([1,2,3])
reverse_ll(ll)
```

3. Write a function to detect a cycle in a singly linked list.
Return True if a cycle exists, else False.

    Input: 1 -> 2 -> 3 -> 4 -> 2 (cycle back to node 2)  
    Output: True


```python
def if_cycle(ll):
    slow = fast = ll.head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True

    return False
        
```

4. Write a function to remove the nth node from the end of a singly linked list.

Input: 1 -> 2 -> 3 -> 4 -> 5, n = 2  
Output: 1 -> 2 -> 3 -> 5  # removed 4 (2nd from end)


```python
def remove_nth_node_fromEnd(ll, n):
    dummy = Node(0)
    dummy.next = ll.head
    ll.head = dummy
    slow = fast = ll.head

    for _ in range(n+1):
        fast = fast.next

    while fast:
        slow = slow.next
        fast = fast.next

    slow.next = slow.next.next
    ll.head = dummy.next
    
    return ll.head

ll = LinkedList([1,2,3,4,5])
remove_nth_node_fromEnd(ll, 2)
ll
```

2. You are given an array of integers.
Write a function to move all zeros to the end 
while maintaining the relative order of non-zero elements.

Input: [0, 1, 0, 3, 12]

Output: [1, 3, 12, 0, 0]


```python
def move_zeroes(arr):
    pos = 0
    for i in range(len(arr)):
        if arr[i] != 0:
            arr[pos], arr[i] = arr[i], arr[pos]
            pos += 1

    return arr

move_zeroes([0, 1, 0, 3, 12])
```

3. Given an array of integers, write a function to find the subarray with the maximum sum (Kadane’s Algorithm).

Input: [-2, 1, -3, 4, -1, 2, 1, -5, 4]  
Output: 6  # max sum subarray: [4, -1, 2, 1]


```python
def max_sum_subarray(arr):
    best_here = arr[0]
    best_all = float('-inf')
    for i in range(1, len(arr)):
        best_here = max(arr[i], best_here+arr[i])
        best_all = max(best_here, best_all)
    return best_all


def max_subarray(arr):
    best_here = arr[0]
    best_all = arr[0]

    start = 0
    best_start = 0
    best_end = 0
    
    for i in range(1, len(arr)):
        if arr[i] > best_here + arr[i]:
            best_here = arr[i]
            start = i
        else:
            best_here = best_here + arr[i]

        if best_here > best_all:
            best_all = best_here
            best_start = start
            best_end = i

    return arr[start : best_end + 1]


sum(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4])) == max_sum_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4])
        
```

```python
def duplicate_elements(arr):
    count = {} 
    for i in arr:
        count[i] = count.get(i, 0) + 1
    return [digit for digit, count in count.items() if count > 1]


def duplicate_elemnts_1_to_len(arr):
    dups = []

    for i in arr:
        idx = abs(i) - 1
        if arr[idx] < 0:
            dups.append(abs(i))
        else:
            arr[idx] = -arr[idx]
    return dups
    

duplicate_elements([4,3,2,7,8,2,3,1])
duplicate_elemnts_1_to_len([4,3,2,7,8,2,3,1])
```

1. Write a function to count the number of vowels (a, e, i, o, u) in a given string.

Input: "hello world"  
Output: 3


```python
def count_vowels(s):
    vow = {'a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U'}
    count = 0
    for l in s:
        if l in vow:
            count += 1
    return count

count_vowels("hello world")
```

```python
def is_anagram(s1, s2):
    if len(s1) != len(s2):
        return False

    from collections import Counter
    count = Counter(s1)
    for l in s2:
        if count[l] < 1:
            return False
        count[l] -= 1
    return all(value == 0 for value in count.values())

is_anagram("listen", "silent")
```

3. Write a function to find the first non-repeating character in a string.
Return the character itself; if none exists, return 'None'.

Input: "swiss"

Output: "w"


```python
def first_non_duplicate(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1

    for char in s:
        if freq[char] == 1:
            return char

    return None

first_non_duplicate("swiss")
```

4. Longest Substring Without Repeating Characters

👉 Problem:
Given a string s, find the length of the longest substring that has no repeating characters.

Input: s = "abcabcbb"

Output: 3

Explanation: "abc" is the longest substring without repeating characters.


```python
def length_of_longest_substring(s):
    max_len = 1
    for i in range(len(s)):
        for j in range(1, len(s)):
            substr = s[i : j+1]

            if len(substr) == len(set(substr)):
                max_len = max(max_len, len(substr))

    return max_len

length_of_longest_substring("abcabcbb")
```

1. Write a recursive function to compute the factorial of a non-negative integer n.

Input: 5  
Output: 120


```python
def factorial(num):
    if num < 0 or not isinstance(num, int):
        raise ValueError("factorial() expects only positive integers")
    
    if num <= 1:
        return 1

    return num * (factorial(num -1))

factorial(5)
```

2. Write a recursive function to find the sum of all elements in an array.

Input: [1, 2, 3, 4]

Output: 10


```python
def sum_arr(arr, i=0):
    if i == len(arr):
        return 0
    return arr[i] + sum_arr(arr, i+1)

sum_arr([1,2,3,5])
```

3. Write a recursive function to find the nth Fibonacci number.

Input: n = 6  
Output: 8   # (0, 1, 1, 2, 3, 5, 8)

```python
def nth_fibonacci(n):

    if n == 0: 
        return 0
    if n == 1:
        return 1

    return nth_fibonacci(n-1) + nth_fibonacci(n-2)

nth_fibonacci(6)
```

4. Implement Binary search using recursion

```python
def binary_search_recursion(arr, target, left = 0, right = None ):
    if right is None:
        right = len(arr) - 1
        
    if left > right:
        return -1

    mid = (left + right) // 2
    if arr[mid] == target:
        return mid

    elif arr[mid] > target :
        binary_search_recursion(arr, target, left, mid - 1)

    else:
        binary_search_recursion(arr, target, mid + 1, right)


binary_search_recursion([1,2,3,4], 2)
```
