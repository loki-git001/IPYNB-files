```python
import speech_recognition as sr

r = sr.Recognizer()

def listen():
    try:
        with sr.Microphone() as source:
            print("Listening...")

            r.adjust_for_ambient_noise(source)
            audio = r.listen(source)
            text = r.recognize_google(audio)
            text = text.lower()
            print("You said:", text)
            return text
        
    except sr.UnknownValueError:
        print("Could not understand audio") 
        return

while True:
    try:
        text = listen()
        if text is None:
            continue
        if "exit" in text:
            print("Exiting...")
            break
    
    except KeyboardInterrupt:
        print("Keyboard Interrupt - Exiting")
        break
```

```python
from gtts import gTTS
from IPython.display import Audio, display
import io


def speak(text):
    tts = gTTS(text=text, lang="it")

    audio_data = io.BytesIO()
    tts.write_to_fp(audio_data)
    audio_data.seek(0)

    display(Audio(audio_data.read(), autoplay=True))


speak("Hello! How are you today?")
```

```python
while True:
    try:
        user_input = listen()

        if user_input is None:
            continue

        if "stop" in user_input or "quit" in user_input:
            speak("Goodbye! Shutting down.")
            break
        elif "hello" in user_input:
            response = "Hello there! How are you doing today?"
        elif "your name" in user_input:
            response = "I am a simple Python speech pipeline."
        else:
            response = f"You just said: {user_input}"

        speak(response)

    except KeyboardInterrupt:
            print("Keyboard Interrupt - Exiting")
            break
```
